## Inspect Remaining Tables

```python
remaining_sheets = [
    "Reservations",
    "Events",
    "EventsLocations",
    "CustomersEvents",
    "CustomersDishes"
]

summary = []

for sheet in remaining_sheets: # Repeat the code once for each sheet name.
    df = pd.read_excel(excel_path, sheet_name=sheet) # Loads the current sheet.

    summary.append({
        "Table": sheet,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "MissingValues": df.isnull().sum().sum(),
        "DuplicateRows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary) # Show it as table.
summary_df
```
